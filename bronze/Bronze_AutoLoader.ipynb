```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
dbutils.widgets.text('folder_name','')
```

```python
folder_name=dbutils.widgets.get('folder_name')
```

```python
df=spark.readStream.format('cloudFiles').option('cloudFiles.format','csv')\
    .option('cloudFiles.schemaLocation',f'abfss://bronze@adventwork.dfs.core.windows.net/checkpoint_{folder_name}')\
    .load(f'abfss://source@adventwork.dfs.core.windows.net/{folder_name}')
```

```python
df.writeStream.format('delta')\
    .outputMode('append')\
    .option('checkpointLocation',f'abfss://bronze@adventwork.dfs.core.windows.net/checkpoint_{folder_name}')\
    .option('path',f'abfss://bronze@adventwork.dfs.core.windows.net/{folder_name}')\
    .trigger(once = True)\
    .start()
```
