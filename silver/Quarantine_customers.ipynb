```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

### **Data reading**

```python
df_invalid = spark.read.format('delta')\
    .load('abfss://silver@adventwork.dfs.core.windows.net/quarantine/customers')
```

### **Checking the accuracy of  Prefix , Marital Status ,Gender , Occupation, TotalChildren , EducationLevel  and HomeOwner columns**

```python
df_invalid.filter(~col("Prefix").isin(['Mr.','Mrs.'])).display()
```

```python
df_invalid.filter(~col('MaritalStatus').isin(['M','S'])).display()
```

```python
df_invalid.filter(~col("Gender").isin(['M','F'])).display()
```

```python
df_invalid.filter(~col("Occupation").isin(['Professional','Management','Skilled Manual','Clerical','Manual'])).display()
```

```python
df_invalid.filter(col("TotalChildren") < 0).display()
```

```python
df_invalid.filter(~col("EducationLevel").isin(['Bachelors','Partial College','High School','Partial High School','Graduate Degree'])).display()
```
