```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
from pyspark import pipelines as dp
```

```python
@dp.table()
def gold_subcategories_stage():
    return spark.read.table('adventure_works.silver.productsubcategories')
```

```python
@dp.materialized_view()
def subcategories_matt_view():
    return spark.read.table('Live.gold_subcategories_stage')
```

```python
dp.create_streaming_table('dim_productsubcategories')
```

```python
dp.create_auto_cdc_flow(
    target='dim_productsubcategories',
    source='Live.subcategories_matt_view',
    keys=['ProductSubcategoryKey'],
    sequence_by=col('SubcategoryName'),
    stored_as_scd_type=1
)
```
