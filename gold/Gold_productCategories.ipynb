```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
from pyspark import pipelines as dp
```

```python
@dp.table()
def gold_categories_stage():
    return spark.read.table('adventure_works.silver.productcategories')
```

```python
@dp.materialized_view()
def categories_matt_view():
    return spark.read.table('Live.gold_categories_stage')
```

```python
dp.create_streaming_table('dim_productcategories')
```

```python
dp.create_auto_cdc_flow(
    target='dim_productcategories',
    source='Live.categories_matt_view',
    keys=['ProductCategoryKey'],
    sequence_by=col('CategoryName'),
    stored_as_scd_type=1
)
```
