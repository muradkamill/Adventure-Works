```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
from pyspark import pipelines as dp
```

```python
@dp.table()
def gold_customers_stage():
    return spark.read.table('adventure_works.silver.customers')
```

```python
@dp.materialized_view()
def customers_matt_view():
    return spark.read.table('Live.gold_customers_stage')
```

```python
dp.create_streaming_table('dim_customers')
```

```python
dp.create_auto_cdc_flow(
    target = 'dim_customers' ,
    source = 'Live.customers_matt_view',
    keys = ['CustomerKey'],
    sequence_by = col('FirstName'),
    stored_as_scd_type = 2

)
```
