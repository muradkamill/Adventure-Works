```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
from pyspark import pipelines as dp
```

```python
@dp.table()
def gold_sales_stage():
    return spark.read.table('adventure_works.silver.sales')
```

```python
@dp.materialized_view()
def sales_matt_view():
    return spark.read.table('Live.gold_sales_stage')
```

```python
dp.create_streaming_table('fact_sales')
```

```python
dp.create_auto_cdc_flow(
    target = 'fact_sales',
    source ='Live.gold_sales_stage',
    keys = ['OrderDate','StockDate','OrderNumber','ProductKey','CustomerKey','TerritoryKey','OrderLineItem','OrderQuantity'],
    sequence_by = col('StockDate') ,
    stored_as_scd_type = 2 
)
```
