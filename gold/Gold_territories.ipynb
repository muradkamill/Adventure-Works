```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
from pyspark import pipelines as dp
```

```python
@dp.table()
def gold_territories_stage():
    return spark.read.table('adventure_works.silver.territories')
```

```python
@dp.materialized_view()
def territories_matt_view():
    return spark.read.table('Live.gold_territories_stage')
```

```python
dp.create_streaming_table('dim_territories')
```

```python
dp.create_auto_cdc_flow(
    target = 'dim_territories',
    source = 'Live.territories_matt_view',
    keys = ['SalesTerritoryKey'],
    sequence_by = col('Region'),
    stored_as_scd_type = 1
)
```
