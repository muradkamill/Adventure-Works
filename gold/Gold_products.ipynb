```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
from pyspark import pipelines as dp
```

```python
@dp.table()
def gold_products_stage():
    return spark.read.table('adventure_works.silver.products')
```

```python
@dp.materialized_view()
def products_matt_view():
    return spark.read.table('Live.gold_products_stage')
    
```

```python
dp.create_streaming_table('dim_products')
```

```python
dp.create_auto_cdc_flow(
    target = 'dim_products',
    source = 'Live.products_matt_view',
    keys = ['ProductKey'],
    sequence_by = col('ProductName'),
    stored_as_scd_type = 2

)
```
